```python
import torch
from torch import nn
import torch.nn.functional as F

X=torch.arange(4)
torch.save(X,'X-file1')
x2=torch.load('X-file1')
x2
```

```python
y=torch.zeros(4)
torch.save([X,y],'x-file2')
x2,y2=torch.load('x-file2')
(x2,y2)
```

```python
class MLP(nn.Module):
    def __init__(self):
        super().__init__()
        self.hidden=nn.Linear(20,128)
        self.out=nn.Linear(128,10)
    def forward(self,X):
        return self.out(F.relu(self.hidden(X)))

X=torch.randn(size=(2,20))
net=MLP()
y=net(X)
```

```python
torch.save(net.state_dict(),'MLP.params')
```

```python
clone=MLP()
clone.load_state_dict(torch.load('MLP.params'))
clone.eval()
```

```python
Y_clone=clone(X)
Y_clone==y
```
